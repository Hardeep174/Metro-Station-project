```python
import os
import pandas as pd

os.makedirs("../data/raw", exist_ok=True)

url = "https://data.ny.gov/resource/wujg-7c2s.csv?$limit=500000"

df = pd.read_csv(url)

print("Shape of dataset:", df.shape)
print(df.head())

df.to_csv("../data/raw/mta_subway_sample.csv", index=False)

print("Sample file saved successfully in ../data/raw/")
```
