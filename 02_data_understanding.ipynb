```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("../data/raw/mta_subway_sample.csv")

# Show basic info
print("Shape of dataset:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nBasic statistics:")
display(df.describe(include='all'))
```

```python
# Convert transit_timestamp to datetime
df["transit_timestamp"] = pd.to_datetime(df["transit_timestamp"], errors="coerce")

print("Updated data types:\n")
print(df.dtypes)

print("\nDate range:")
print("Start:", df["transit_timestamp"].min())
print("End:", df["transit_timestamp"].max())
```

```python
# Check unique values in important columns
print("Unique transit modes:", df["transit_mode"].unique())
print("Number of unique stations:", df["station_complex"].nunique())
print("Unique boroughs:", df["borough"].unique())
print("Unique payment methods:", df["payment_method"].unique())

print("\nTarget variable summary: ridership")
print(df["ridership"].describe())
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Create time-based features
df["hour"] = df["transit_timestamp"].dt.hour
df["day"] = df["transit_timestamp"].dt.day
df["month"] = df["transit_timestamp"].dt.month
df["day_of_week"] = df["transit_timestamp"].dt.dayofweek
df["is_weekend"] = df["day_of_week"].apply(lambda x: 1 if x >= 5 else 0)

print(df[["transit_timestamp", "hour", "day", "month", "day_of_week", "is_weekend"]].head())
```

```python
# Save updated file for next step
df.to_csv("../data/processed/mta_subway_understood.csv", index=False)

print("Processed preview file saved in ../data/processed/")
```
