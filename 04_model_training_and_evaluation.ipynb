```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
# Load train-test data
X_train = pd.read_csv("../data/processed/X_train.csv")
X_test = pd.read_csv("../data/processed/X_test.csv")
y_train = pd.read_csv("../data/processed/y_train.csv").squeeze()
y_test = pd.read_csv("../data/processed/y_test.csv").squeeze()

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
# Create models
lr = LinearRegression()
dt = DecisionTreeRegressor(random_state=42)
rf = RandomForestRegressor(n_estimators=100, random_state=42, n_jobs=-1)
```

```python
# Train models
lr.fit(X_train, y_train)
dt.fit(X_train, y_train)
rf.fit(X_train, y_train)

print("All models trained successfully!")
```

```python
# Predictions
lr_pred = lr.predict(X_test)
dt_pred = dt.predict(X_test)
rf_pred = rf.predict(X_test)
```

```python
def evaluate_model(name, y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    
    print(f"\n{name} Results")
    print("-" * 30)
    print("MAE :", mae)
    print("RMSE:", rmse)
    print("R²  :", r2)
    
    return [name, mae, rmse, r2]
```

```python
lr_results = evaluate_model("Linear Regression", y_test, lr_pred)
dt_results = evaluate_model("Decision Tree", y_test, dt_pred)
rf_results = evaluate_model("Random Forest", y_test, rf_pred)
```

```python
results_df = pd.DataFrame(
    [lr_results, dt_results, rf_results],
    columns=["Model", "MAE", "RMSE", "R2 Score"]
)

print(results_df)
```

```python
best_model = results_df.sort_values(by="R2 Score", ascending=False).iloc[0]
print("Best model:")
print(best_model)
```

```python
import os
import joblib

os.makedirs("../models", exist_ok=True)

joblib.dump(lr, "../models/linear_regression_model.pkl")
joblib.dump(dt, "../models/decision_tree_model.pkl")
joblib.dump(rf, "../models/random_forest_model.pkl")

print("All models saved successfully in ../models/")
```

```python
results_df.to_csv("../data/processed/model_results.csv", index=False)
print("Model results saved successfully!")
```
